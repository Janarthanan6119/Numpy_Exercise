## Array creation

### Create a numpy array of size 10, filled with zeros.

```python
# your code goes here
print(np.zeros(10,dtype=int))
# np.array([0]*10)

'''
Difference between dtype=int ,dtype=np.int 

dtype=np.int - means in older times,they used np.int for integer data type.Currently,np.int is removed from the NUmpy Library
So,we must use dtype=int
'''

'''
Difference between dtype=int32 and dtype=int64

Main difference is,elements will store in 32 bits and 64 bits

32 bits means - Each element will take 4 bytes(Min - -2^31, Max - 2^31 - 1)
64 bits means - Each Element will take 8 bytes(Min - -2^63, Max - 2^63 - 1)
'''
```
