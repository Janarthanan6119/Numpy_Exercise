### Create a numpy array of size 10, filled with zeros.

```python
# your code goes here
print(np.zeros(10,dtype=int))
# np.array([0]*10)

'''
Difference between dtype=int ,dtype=np.int 

dtype=np.int - means in older times,they used np.int for integer data type.Currently,np.int is removed from the NUmpy Library
So,we must use dtype=int
'''

'''
Difference between dtype=int32 and dtype=int64

Main difference is,elements will store in 32 bits and 64 bits

32 bits means - Each element will take 4 bytes(Min - -2^31, Max - 2^31 - 1)
64 bits means - Each Element will take 8 bytes(Min - -2^63, Max - 2^63 - 1)
'''
```

Boolean arrays (also called masks)

Given the X numpy array, make a mask showing negative elements

```python
X = np.array([-1,2,5,-9,-6,0,-3])
temp=X<0
temp
```

Given the X numpy array, get the negative elements

```python
X = np.array([-1,2,5,-9,-6,0,-3])
temp = X<0
X[temp]
```

Given the X numpy array, get numbers higher than 5

```python
X = np.array([12,2,5,9,8,7])
temp = X>5
X[temp]
```

Given the X numpy array, get numbers higher than the elements mean

```python
X=np.array([12,2,5,9,8,7])
print(X.mean())
temp = X > X.mean()
X[temp]
```

Given the X numpy array, get numbers equal to 2 or 10

```python
X = np.array([12,2,5,9,8,7,10,2])
temp = (X==2) | (X==10)
X[temp]
'''
and, or, not → ordinary Python Boolean values(which will check the Single boolean value(i.e. it will produce either true or false))
&, |, ~ → element-wise NumPy Boolean operations(which will check element-by-element in Arrays)
'''
```
